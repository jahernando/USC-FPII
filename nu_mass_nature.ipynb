## 1. Experimental Limits on the Neutrino Mass

### Neutrino mass from oscillations: recap

Oscillation experiments measure mass-squared differences:

$$\Delta m^2_{21} = m_2^2 - m_1^2 = (7.41^{+0.21}_{-0.20})\times 10^{-5}\,\mathrm{eV}^2$$

$$|\Delta m^2_{31}| = (2.511^{+0.028}_{-0.027})\times 10^{-3}\,\mathrm{eV}^2 \quad (\text{NuFit-6.0, 2024})$$

These fix the *splittings* but not the absolute scale $m_1$. Two orderings are allowed:

| Ordering | $m_1$ | $m_2$ | $m_3$ | $\sum m_i$ (min) |
|---|---|---|---|---|
| **Normal (NH)** | lightest | $m_1+7\times10^{-3}$ eV | $m_1+50\times10^{-3}$ eV | $\geq 57$ meV |
| **Inverted (IH)** | $m_3+50\times10^{-3}$ eV| $m_3+51\times10^{-3}$ eV | lightest | $\geq 100$ meV |

**Current status on ordering**: 
  - NuFit-6.0 prefers NH at $\sim2.5\sigma$; 
  - JUNO first results (2025) favour NH at $\sim2\sigma$ from reactor data alone.
  - A definitive measurement requires JUNO full statistics, DUNE, or HyperK.

### Kinematic neutrino mass measurements

Beta-decay endpoint spectroscopy provides a **model-independent** measurement of the neutrino mass. It probes the *effective electron antineutrino mass*:

$$m^2_\nu \equiv \sum_i |U_{ei}|^2\, m_i^2$$

This quantity is independent of the Majorana/Dirac nature, CP phases, and the cosmological model.

Near the endpoint $E_0$ of the $\beta$-decay energy spectrum, the creation of (anti)neutrinos modifies the available phase space. Each mass eigenstate contributes with its own admixture but is typically not resolved; the shape distortion is maximal near $E_0$:

$$
\frac{dN}{dT} \propto (E_0 - T) \sqrt{(E_0 - T)^2 - m^2_\nu} \;\Theta(E_0 - T - m_\nu)
$$

### KATRIN experiment

The [KATRIN experiment](http://www.katrin.kit.edu) (KArlsruhe TRItium Neutrino) at KIT (Karlsruhe, Germany) performs precision spectroscopy of the molecular tritium $\beta$-decay close to the kinematic endpoint $E_0 \approx 18.6$ keV:

$$\mathrm{T}_2 \to {}^3\mathrm{HeT}^+ + e^- + \bar{\nu}_e$$

### Cosmological bounds on neutrino mass

Massive neutrinos suppress small-scale structure formation (free-streaming). Cosmological observations constrain $\Sigma m_\nu = m_1 + m_2 + m_3$.

#### CMB + BAO: Planck 2018

The Planck Collaboration (2018) combining CMB temperature/polarisation and lensing with BAO data (BOSS, 6dFGS, SDSS) obtains [[P18]](https://arxiv.org/abs/1807.06209):

$$\Sigma m_\nu < 0.12 \text{ eV} \quad (95\%\,\mathrm{CL})$$

This is the most stringent laboratory-independent constraint on the neutrino mass scale. It disfavours the IH quasi-degenerate region ($\Sigma m_\nu \gtrsim 0.10$ eV) but is still compatible with both orderings in the non-degenerate limit.

#### DESI 2024: Baryon Acoustic Oscillations

The **Dark Energy Spectroscopic Instrument (DESI)** measured BAO from >6 million extragalactic objects (galaxies, quasars, Ly$\alpha$ forest) across $0.1 < z < 4.2$ in its first year of observations [[D24]](https://arxiv.org/abs/2404.03002).

Combined with CMB (Planck + ACT lensing), in flat $\Lambda$CDM + $\sum m_\nu$:

$$\Sigma m_\nu < 0.072 \text{ eV} \quad (95\%\,\mathrm{CL}) \quad [\text{DESI BAO + CMB}]$$

## 2. How to Provide Mass to the Neutrinos

### Dirac neutrino masses

A Dirac spinor fulfills the Dirac equation and can be decomposed into left and right chirality parts:

$$
\Psi = \Psi_L + \Psi_R
$$

It admits a global phase redefinition $\Psi' = e^{i\theta}\Psi$ without physical consequences. This global symmetry is associated with a conserved charge (electric charge, lepton number, etc.).

A mass term in the Lagrangian has the form:

$$
-m \bar{\Psi} \Psi = -m (\bar{\Psi}_L \Psi_R + \bar{\Psi}_R \Psi_L)
$$

It couples both chiralities — **a mass term requires both left and right components.**

### Dirac masses in the SM

The SM places left-chiral fields into doublets and right-chiral fields into singlets of $SU(2)_L$. Bare masses are forbidden by gauge invariance; they arise after SSB via Yukawa couplings with the Higgs.

For one lepton family:

$$
L = \begin{pmatrix} \nu_L \\ l_L \end{pmatrix}, \quad l_R, \quad
\Phi = \begin{pmatrix} \phi^+ \\ \phi^0 \end{pmatrix}
\xrightarrow{\text{SSB}}
\langle \Phi \rangle = \begin{pmatrix} 0 \\ v/\sqrt{2} \end{pmatrix}
$$

The charged-lepton mass arises from the Yukawa coupling $\lambda_e \bar{L} \Phi \, e_R + \text{h.c.}$, giving $m_e = \lambda_e \, v/\sqrt{2}$ after SSB.

But neutrinos remain massless in this construction — there is no right-chiral neutrino field to couple to.

To give mass to the neutrino, we must introduce a **sterile right-chiral field** $\nu_R$, a singlet under the SM gauge group, and couple it via the Higgs charge conjugate:

$$
\tilde{\Phi} = i\sigma^2 \Phi^* \xrightarrow{\text{SSB}}
\langle \tilde{\Phi} \rangle = \begin{pmatrix} v/\sqrt{2} \\ 0 \end{pmatrix}
$$

The new Yukawa interaction is:

$$
-\mathcal{L}_{\mathrm{mass},\,\nu} = \lambda_\nu \, \bar{L} \, \tilde{\Phi} \, \nu_R + \mathrm{h.c.}
$$

After SSB this gives a Dirac neutrino mass:

$$
m_\nu = \lambda_\nu \frac{v}{\sqrt{2}}
$$

The uncomfortable question arises: *why are neutrino masses at least 6 orders of magnitude smaller than their charged-lepton partners?*

$$
\lambda_\nu \ll \lambda_l
$$

This enormous hierarchy of Yukawa couplings has no explanation within the SM — it is one of the strongest motivations for considering the Majorana alternative.

### Majorana fermions

[Ettore Majorana](https://en.wikipedia.org/wiki/Ettore_Majorana) in the 1930s proposed that neutral fermions can be *self-conjugate*.

| |
| :--: |
| <img src="./imgs/numass_majorana_abstract.png" width = 400> |

Majorana introduced a representation where all $\gamma^\mu$ are pure imaginary:

$$
\gamma^0 = \begin{pmatrix} 0 & \sigma^2 \\ \sigma^2 & 0\end{pmatrix}, \;
\gamma^1 = \begin{pmatrix} i\sigma^3 & 0 \\ 0 & i\sigma^3\end{pmatrix}, \;
\gamma^2 = \begin{pmatrix} 0 & -\sigma^2 \\ \sigma^2 & 0\end{pmatrix}, \;
\gamma^3 = \begin{pmatrix} i\sigma^1 & 0 \\ 0 & i\sigma^1 \end{pmatrix}
$$

In this representation, the Dirac equation is real and admits a real solution: $\Psi(x) = \Psi^*(x)$. A particle is its own antiparticle.

#### The Majorana condition

In any representation of the Dirac matrices, the Majorana condition translates to:

$$
\Psi = \Psi^c, \quad \text{where} \quad \Psi^c \equiv C \bar{\Psi}^T = C \gamma^0 \Psi^*
$$

The charge conjugation matrix $C$ can be obtained by imposing covariance of the Dirac equation under charge conjugation.

In the Weyl-chiral representation:

$$
\Psi = \begin{pmatrix} \psi_L \\ \psi_R \end{pmatrix} \; \to \;
\Psi^c = i\gamma^2 \Psi^* = \begin{pmatrix} 0 & i\sigma^2 \\ -i\sigma^2 & 0 \end{pmatrix} \begin{pmatrix} \psi_L^* \\ \psi_R^* \end{pmatrix} = \begin{pmatrix} i\sigma^2 \psi_R^* \\ -i\sigma^2 \psi_L^* \end{pmatrix}
$$

The charge conjugate of a left-chiral spinor is right-chiral:

$$
\Psi_L = \begin{pmatrix} \psi_L \\ 0 \end{pmatrix} \to
(\Psi_L)^c = \begin{pmatrix} 0 \\ -i \sigma^2 \psi^*_L \end{pmatrix}
$$

#### Constructing the Majorana spinor
A Majorana spinor can be constructed from a single left Weyl spinor $\psi_L$:

$$
\Psi^M = \Psi + \Psi^c = \begin{pmatrix} \psi_L \\ 0 \end{pmatrix} + \begin{pmatrix} 0 \\ -i \sigma^2 \psi^*_L \end{pmatrix} = \begin{pmatrix} \psi_L \\ -i \sigma^2 \psi^*_L \end{pmatrix}
$$

This shows that a Majorana fermion:
 - Has left **and** right chiral components — therefore it **has mass** (otherwise we recover the Weyl equation).
 - Has **no global phase symmetry**: $\psi_L \to e^{i\phi} \psi_L$ transforms the upper and lower components with opposite phases — therefore **no conserved charge**. It is a purely neutral fermion.

Neutrinos have mass — as demonstrated by oscillations — and they are the only fundamental fermions that could be Majorana particles.

*What, then, is the nature of neutrinos: Dirac or Majorana?*

### The Weinberg operator

We have seen how to construct a Majorana fermion from a single left-chiral field $\psi_L$. The question now is: how do we generate a Majorana mass while respecting the gauge symmetry of the SM?

The mass term for left-chiral Majorana neutrinos is:

$$
-\frac{m}{2} (\overline{\nu_L} \nu^c_L + \overline{\nu^c_L} \nu_L)
$$

Since $\nu_L$ is part of an $SU(2)_L$ doublet, gauge invariance requires a *double* coupling with the Higgs — the 5-dimensional **Weinberg operator** [Wein]:

$$
\mathcal{L}_W = -\frac{\alpha}{\Lambda} (\bar{L}\tilde{\Phi})(\tilde{\Phi}^T L^c) + \mathrm{h.c.}
$$

where $\alpha$ is a dimensionless coupling and $\Lambda$ is an unknown large energy scale.

After SSB:

$$
\mathcal{L}_W = -\frac{\alpha v^2}{2\Lambda}(\overline{\nu_L} \nu^c_L + \overline{\nu^c_L} \nu_L) \quad \Rightarrow \quad m_\nu = \frac{\alpha v^2}{\Lambda}
$$

For $\alpha \sim 1$, $v = 246$ GeV, and $m_\nu \sim 1$ eV: $\Lambda \sim 10^{14}$ GeV.


The Weinberg operator is the **only 5-dimensional operator** in the SM effective field theory expansion — it is the first signal of physics beyond the SM at low energy.

### Mixing and the $U_{PMNS}$ Matrix

With three families, the Yukawa couplings become $3\times3$ complex matrices.

The coupling with the Higgs allows Nature to relate flavour eigenstates to mass eigenstates via a unitary matrix $U$.

This is the matrix responsible, among other things, for neutrino oscillations.

The derivation depends on the neutrino nature: whether it is a Dirac or a Majorana fermion.

### Dirac masses: three families

For three families, the Yukawa interaction mixes flavours:

$$
-\mathcal{L}_H = \sum_{i,j=1}^3 \left(\lambda^l_{ij} \bar{L}'_i \Phi \, l'_{jR} + \lambda^\nu_{ij} \bar{L}'_i \tilde{\Phi} \, \nu'_{jR}\right) + \mathrm{h.c.}
$$

After SSB, the complex matrices $\frac{v}{\sqrt{2}}\lambda_\alpha$ are diagonalised by two unitary transformations:

$$
\frac{v}{\sqrt{2}} \lambda_\alpha = U^\dagger_\alpha \, M_\alpha V_\alpha
$$

relating weak and mass eigenstates: $l_L = U_l \, l'_L$, $\nu_L = U_\nu \, \nu'_L$.

The NC is **unaffected** (unitarity: $U^\dagger U = I$), but the CC acquires the PMNS matrix:

$$
-\mathcal{L}_{CC} = \frac{g}{\sqrt{2}} \bar{l}_L \gamma^\mu (U^\dagger_l U_\nu) \nu_L W^+_\mu + \mathrm{h.c.}
$$

In the quark sector, the analogous matrix is the $V_{CKM}$ (Cabibbo-Kobayashi-Maskawa).

### The $U_{PMNS}$ matrix (Pontecorvo–Maki–Nakagawa–Sakata) [MNS, Pon]

$$
U_{PMNS} = U^\dagger_l \, U_\nu
$$

The flavour and mass eigenstates are related by: $\nu'_\alpha = \sum_k U_{\alpha k} \, \nu_k$

**Dirac case**: $U_{PMNS}$ has 3 angles and **1 CP phase** ($\delta_{CP}$). We can reabsorb 5 phases using the global symmetries of the 6 Dirac fields.

### Majorana masses: three families

In the Majorana case, we extend the SM with the Weinberg operator for three families:

$$
-\mathcal{L}_m = \sum_{\beta, \beta'} \frac{\alpha^\nu_{\beta\beta'}}{\Lambda} (\bar{L}'_\beta \tilde{\Phi})(\tilde{\Phi}^T L'^c_{\beta'}) + \mathrm{h.c.}
$$

After SSB, the $3\times3$ symmetric complex matrix $\frac{v}{\sqrt{2}\Lambda}\alpha_\nu$ is diagonalised by a single unitary transformation.

The resulting mass matrix is symmetric — a hallmark of the Majorana nature.


**Majorana case**: Majorana fields have no global phase freedom, so only 3 phases (from charged leptons) can be reabsorbed. The $U_{PMNS}$ matrix has 3 angles and **3 phases**: $\delta_{CP}$ and two Majorana phases $\eta_1, \eta_2$.

The Majorana phases $\eta_{1,2}$ have no effect on oscillation probabilities — they can only be probed through lepton-number-violating processes such as $\beta\beta0\nu$.

### The seesaw mechanism

The Weinberg operator is a low-energy effective description. At high energies it is *UV-completed* by the **seesaw mechanism**, which naturally explains the smallness of neutrino masses.

It implies the existence of new particles mediating the interaction at tree level.

Three main realisations:

| Type | New particle | Mass scale | Key feature |
|---|---|---|---|
| **Type I** | Heavy right-handed neutrino $N_R$ | $M_R \sim 10^{14}$ GeV | Minimal, most studied |
| **Type II** | Scalar triplet $\Delta$ | $M_\Delta \sim 10^{13}$ GeV | Predicts same-sign dileptons |
| **Type III** | Fermion triplet $\Sigma$ | $M_\Sigma \sim 10^{13}$ GeV | Testable at colliders |

#### Type I seesaw

Extend the SM with $n$ heavy right-handed (sterile) neutrinos $N_{Ri}$, singlets under the SM gauge group.

The mass Lagrangian (one generation) is:

$$-\mathcal{L} = y_\nu\,\bar{L}\,\tilde{\Phi}\,N_R + \frac{1}{2}M_R\,\overline{N^c_R}\,N_R + \text{h.c.}$$

After electroweak symmetry breaking ($\langle\Phi\rangle = v/\sqrt{2}$), the Dirac mass $m_D = y_\nu v/\sqrt{2}$ and the Majorana mass $M_R \gg m_D$ give the mass matrix:

$$
\begin{pmatrix} \nu_L \\ N_R \end{pmatrix} \to
\mathcal{M} = \begin{pmatrix} 0 & m_D \\ m_D & M_R \end{pmatrix}
$$

Diagonalising for $m_D \ll M_R$ yields two eigenvalues:

$$m_\nu \approx \frac{m_D^2}{M_R} \quad (\text{light}), \qquad M_N \approx M_R \quad (\text{heavy})$$

The larger $M_R$, the smaller $m_\nu$ — hence the name *seesaw*.

The heavy neutrinos $N_R$ could have played a major role in the asymmetry of the Universe via **leptogenesis**!

## 3. How to Determine the Nature of the Neutrino

### Dirac vs Majorana: experimental distinction

| |
| :--: |
| <img src="./imgs/pidecay_majorana_dirac.png" width = 600> |

If the neutrino is Majorana, a neutrino created in a $\pi^+$ decay would be able to interact and produce either $\mu^+$ or $\mu^-$.

However, the lepton-number-violating rate is suppressed by $(m_\nu/E)^2$:

| | $h = -1$ | $h = +1$ |
| :--:  | :--: | :--: |
| $\nu_L$ | $\sim 1$ | $\mathcal{O}(m/E)$ |
| $\nu_R$ | $\mathcal{O}(m/E)$ | $\sim 1$ |

 * For $m_\nu \sim 1$ eV and $E \sim 1$ MeV: suppression $\sim 10^{-12}$
 * For $m_\nu \sim 1$ eV and $E \sim 1$ GeV: suppression $\sim 10^{-18}$

Direct observation of lepton-number violation in scattering is experimentally impossible. Majorana and Dirac are *experimentally indistinguishable* except through the very rare $\beta\beta0\nu$ process.

### Double beta decay without neutrinos

| |
| :--: |
| <img src="./imgs/bb0nu_Feynman.png" width = 400> |

The only experimentally accessible process is the hypothetical and very rare double beta decay *without neutrinos*:

$$
(A, Z) \to (A, Z+2) + 2e^- + Q_{\beta\beta}
$$

The three massive neutrinos participate, each with a suppressed coefficient $U^2_{ei} \, m_i / E$. The transition amplitude is proportional to the **effective Majorana mass**:

$$
m_{\beta\beta} \equiv \left|\sum_i U^2_{ei} \, m_i\right|
$$

### $\beta\beta 2\nu$: the allowed double beta decay

The $\beta\beta2\nu$ (with neutrinos) was proposed by [M. Goeppert-Mayer](https://en.wikipedia.org/wiki/Maria_Goeppert-Mayer) in 1935:

$$
(A, Z) \to (A, Z+ 2) + 2 e^- + 2 \bar{\nu}_e + Q_{\beta\beta}
$$

This is a second order decay that happens in 35 isotopes where single $\beta$ is kinematically forbidden. In particular in: $^{48}\mathrm{Ca}$, $^{76}\mathrm{Ge}$, $^{82}\mathrm{Se}$, $^{100}\mathrm{Mo}$, $^{130}\mathrm{Te}$ and $^{136}\mathrm{Xe}$.

The half-life is $\mathcal{O}(10^{18}-10^{24})$ yr.

| |
| :--: |
|<img src="./imgs/bb2nu_isotopes_T12.png" width = 500 > |

In 1939 W. Furry proposed the hypothetical decay $\beta\beta0\nu$ if neutrinos are Majorana. The half-life is:

$$
\left(T^{0\nu}_{1/2}\right)^{-1} = G^{0\nu} \, \left| M^{0\nu} \right|^2 \, m^2_{\beta\beta}
$$

where $G^{0\nu}$ is the phase factor and $|M^{0\nu}|^2$ the nuclear matrix element (NME) squared.

The nuclear matrix element has a large theoretical uncertainty. $M^{0\nu} \sim \mathcal{O}(1)$.

### The effective Majorana mass

The Majorana mass $m_{\beta\beta}$ depends on the $U_{\alpha i}$ matrix elements, the mass of the lightest neutrino $m_0$, and the Majorana phases.

In **NH**:

$$
m_{\beta\beta} = \left| m_0 c^2_{12} c^2_{13} + \sqrt{m^2_0 + \Delta m^2_{21}} \, s^2_{12} c^2_{13} e^{2i (\eta_2 -\eta_1)} + \sqrt{m^2_0 + \Delta m^2_{32} + \Delta m^2_{21}} \, s^2_{13} e^{-2i(\delta_{CP} + \eta_1)}\right|
$$

In the limit $m_0 \to 0$: $m_{\beta\beta} \simeq 1.1 - 4.2$ meV (NH), $m_{\beta\beta} \simeq 15 - 50$ meV (IH).

| :--: |
| <img src="./imgs/mbb_vs_mlight.png" width = 400> |

The Majorana mass $m_{\beta\beta}$ vs the lightest neutrino mass for Normal (red) and Inverted (blue) Hierarchy.

The objective of the next generation of $\beta\beta0\nu$ experiments is to cover the IH region.

### Interplay: $m^{eff}_{\nu_e}$, $m_{\beta\beta}$, and $\Sigma m_\nu$

| Quantity | Formula | Probes |
| :-- | :-- | :-- |
| $m^{eff}_{\nu_e}$ | $\sqrt{\sum_i |U_{ei}|^2 m_i^2}$ | absolute mass scale (Dirac or Majorana) |
| $m_{\beta\beta}$ | $|\sum_i U^2_{ei} m_i|$ | Majorana nature + CP phases |
| $\Sigma m_\nu$ | $m_1 + m_2 + m_3$ | cosmological (model-dependent) |

The KATRIN limit $m^{eff}_{\nu_e} < 0.45$ eV constrains $m_{\beta\beta}$:

 * In the **quasi-degenerate regime** ($m_0 \gg \Delta m_{ij}$): $m_{\beta\beta} \lesssim m^{eff}_{\nu_e} < 0.45$ eV
 * For **IH**, the minimum $m_{\beta\beta} \sim 15\text{--}50$ meV is well below KATRIN reach — $\beta\beta0\nu$ experiments are the decisive probe.

## 4. The Neutrinoless Double Beta Decay Search

### $\beta\beta$ experimental signature

| |
| :--: |
|<img src="./imgs/bb2nu_bb0nu_spectrum.png" width = 400> |

- Signature: a mono-energetic spectrum of the two electrons whose total energy is $Q_{\beta\beta}$

The number of expected $\beta\beta0\nu$ events:

$$
N_{\beta\beta} = \ln(2) \frac{N_A}{W} \frac{a \, \epsilon \, M \, t}{T^{0\nu}_{1/2}}
$$

where $N_A$ is Avogadro's number, $W$ the molar mass, $a$ the isotopic abundance, $\epsilon$ the detection efficiency in the RoI, $t$ the time of exposure, $M$ the target mass.

### Background and sensitivity

The signal is identified by:
  - Energy resolution, Region of Interest (RoI), $\Delta E = \mathrm{FWHM} = 2\sqrt{2\ln 2}\;\sigma \simeq 2.355\;\sigma$
  - Other discrimination variables: tracks (2 $\beta$), pulse shape, ...

The background comes from:
  - Cosmogenic muons (spallation) — detector installed underground with veto system. Reduction: $\mathcal{O}(10^6)$
  - Natural radioactivity (U, Th chains) — ultra-radiopure materials. Activities: $\mathcal{O}(10^{-6})$ Bq/kg

| :--: | :--: |
| <img src="./imgs/Th232_chain.png" width = 300 align='left'> | <img src="./imgs/U238_chain.png" width = 300 align='right'> |

| |
| :--: |
| <img src="./imgs/muflux_undergroundlabs.png" width = 400 align='center'> |
| Water equivalent meters of the main underground laboratories |

### Sensitivity regimes

The number of background events: $N_{bkg} \propto b \, \Delta E \, M \, t$

The sensitivity $\mathcal{S}$ has two domains:

**i) Background-free experiment**:
$$\mathcal{S}(T^{0\nu}_{1/2}) \propto \epsilon \, M \, t$$

**ii) With background index $b$**:
$$\mathcal{S}(T^{0\nu}_{1/2}) \propto \frac{N_{sig}}{\sqrt{N_{bkg}}} = \sqrt{\frac{M \, t}{b \, \Delta E}}$$

| |
| :--: |
| <img src="./imgs/Xe_mbb_vs_exposure_bkginRoI.png" width = 400 align='center'> |

Possible limits on $m_{\beta\beta}$ of a perfect Xe experiment for different bkg-index in RoI [[4]](https://arxiv.org/abs/1502.00581).

To cover the IH allowed region we aim for a 1 ton detector, 1 count/(RoI ton yr) — the next-generation experiment holy grail!

### Detector requirements

The main ingredients for a $\beta\beta$ detector are:

   - Large exposure of the target isotope, $M \, t$
   - Excellent energy resolution, small $\Delta E$
   - Ultra-radiopure detector and extra handles to reject background, very small $b$

For next-generation experiments we aim: $M \, t \sim \mathcal{O}(1)$ ton yr, $\Delta E \sim \mathcal{O}(1-0.1)$ %, and $b \sim \mathcal{O}(0.1)$ counts/(RoI ton yr).

**Exercise: Experimental sensitivity as a function of exposure**

The half-life sensitivity of a $\beta\beta_{0\nu}$ search depends on exposure $M \cdot t$,
background index (BI) and energy resolution $\Delta E$:

$$
T^{0\nu}_{1/2}{}^{\rm sens} \propto \begin{cases}
M \cdot t & \text{(zero-background)} \\
\sqrt{M \cdot t / (\mathrm{BI} \cdot \Delta E)} & \text{(background-dominated)}
\end{cases}
$$

## Conclusions

- Neutrino oscillations establish that at least two neutrino masses are non-zero, with $\Sigma m_\nu^{\min}(\mathrm{NH}) \approx 59$ meV.
- KATRIN constrains $m^{\mathrm{eff}}_{\nu_e} < 0.45$ eV; its design goal of $< 0.2$ eV is within reach.
- Cosmological observations (Planck, DESI) push $\Sigma m_\nu < 0.072\text{--}0.12$ eV, approaching the minimum allowed by oscillations.
- The SM cannot explain neutrino masses without new physics: either sterile $\nu_R$ (Dirac) or the Weinberg operator at scale $\Lambda \sim 10^{14}$ GeV (Majorana).
- If neutrinos are Majorana, the seesaw mechanism naturally explains the smallness of their mass and provides a mechanism for the matter-antimatter asymmetry (leptogenesis).
- The only experimental probe of the Majorana nature is $\beta\beta0\nu$. Current best limits: $T^{0\nu}_{1/2} > 3.8 \times 10^{26}$ yr (KamLAND-Zen 800), $T^{0\nu}_{1/2} > 2.8 \times 10^{26}$ yr (LEGEND-200 combined).
- Next-generation experiments (KamLAND2-Zen, LEGEND-1000, nEXO, NEXT-HD) aim to cover the IH band ($m_{\beta\beta} > 15$ meV) within the next decade.
