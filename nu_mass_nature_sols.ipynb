**Exercise: KATRIN — endpoint spectrum and kinematic neutrino mass**

KATRIN measures the tritium $\beta$-decay spectrum near the endpoint $Q_\beta = 18\,574$ eV.
Near the endpoint, the differential rate behaves as:

$$
\frac{dN}{dT} \propto (Q_\beta - T) \sqrt{(Q_\beta - T)^2 - m^2_{\nu_e}} \;\Theta(Q_\beta - T - m_{\nu_e})
$$

The effective mass $m^{eff}_{\nu_e} = \sqrt{\sum_i |U_{ei}|^2 m_i^2}$ sets the kinematic endpoint.

**Exercise: The seesaw mechanism and the PMNS matrix**

**Part A — Seesaw numerics:**

The Type I seesaw gives $m_\nu \approx m_D^2 / M_R$, where $m_D = y_\nu v/\sqrt{2}$.

1. Compute $m_\nu$ for $m_D = m_\tau = 1.777$ GeV and $M_R = 10^{14}$ GeV. Is the result in the right ballpark?
2. What value of $M_R$ would give $m_\nu = 0.05$ eV (the atmospheric scale) if $m_D = m_t = 173$ GeV?
3. The three families have very different $m_D$ (think $m_e, m_\mu, m_\tau$ or $m_u, m_c, m_t$). Does the seesaw naturally predict a neutrino mass hierarchy? Discuss.

**Part B — PMNS phases:**

4. A $3\times3$ unitary matrix has 9 real parameters (3 angles + 6 phases). How many phases are physical in the Dirac case? And in the Majorana case? Explain which field redefinitions are available in each case.
5. Why do the Majorana phases $\eta_1, \eta_2$ not affect oscillation probabilities? Show it explicitly for the two-flavour case.

```python
# Part A — Seesaw numerics

# Q1: m_D = m_tau, M_R = 1e14 GeV
mD_tau = 1.777          # GeV
MR     = 1e14           # GeV
m_nu   = mD_tau**2 / MR # GeV
print(f'Q1:  m_nu = {m_nu*1e9:.3f} eV   (m_D = m_tau = {mD_tau} GeV, M_R = 1e14 GeV)')

# Q2: target m_nu = 0.05 eV = 5e-11 GeV, m_D = m_top = 173 GeV
m_nu_target = 5e-11   # GeV = 0.05 eV
mD_top      = 173.0   # GeV
MR_req      = mD_top**2 / m_nu_target
print(f'Q2:  M_R = {MR_req:.2e} GeV   (m_D = m_top = {mD_top} GeV, m_nu = 0.05 eV)')
```

**Q1.** $m_\nu \approx 3 \times 10^{-8}$ eV — three orders of magnitude smaller than the atmospheric scale. Seesaw with $m_D \sim m_\tau$ and $M_R = 10^{14}$ GeV is *not* enough; one typically needs $m_D \sim v \sim 10^2$ GeV for the right ballpark.

**Q2.** $M_R \approx 6 \times 10^{14}$ GeV — very close to the GUT scale. The seesaw naturally ties neutrino masses to grand-unified physics.

**Q3.** If $m_{D,i} \propto m_i^{\text{up}}$ (top quark pattern), then $m_{\nu_i} \propto m_i^2 / M_R \Rightarrow$ a **very strong** hierarchy $m_{\nu_1} : m_{\nu_2} : m_{\nu_3} \sim (m_u : m_c : m_t)^2$ — too steep. In practice the PMNS mixing is large and the mass spectrum milder, which suggests $M_R$ is *also* a non-trivial matrix and/or a different Dirac structure.

**Part B — PMNS phases:**

**Q4.** Start with 9 parameters: 3 mixing angles + 6 phases.

 * **Dirac case**: the lepton fields $\ell_\alpha, \nu_\alpha$ have $6$ global phases available. One overall $U(1)$ remains ungauged $\Rightarrow$ $6 - 1 = 5$ phases can be reabsorbed. Left with **1 physical phase** ($\delta_{CP}$).

 * **Majorana case**: Majorana fields have no continuous phase symmetry (mass term $\bar{\nu^c} \nu$ is not invariant under $\nu \to e^{i\phi}\nu$). Only the 3 charged-lepton phases can be rotated away, so $6 - 3 = 3$ physical phases: $\delta_{CP} + \eta_1, \eta_2$ (two Majorana phases).

**Q5.** In a 2-flavour Majorana $U$:

$$U = \begin{pmatrix} c & s\,e^{i\eta} \\ -s & c\,e^{i\eta} \end{pmatrix}$$

The oscillation probability depends on $U_{\alpha i} U^*_{\beta i} U^*_{\alpha j} U_{\beta j}$ — bilinears in which Majorana phases $e^{i\eta}$ always appear as $e^{i\eta} e^{-i\eta} = 1$. Majorana phases cancel. They only show up in **LNV processes** like $\beta\beta0\nu$, where amplitudes go as $U_{e i}^2 \propto e^{2i\eta}$ (no cancellation).

$$

3. In IH, does $m_{\beta\beta}$ ever cancel to zero? Why or why not?

**Q2.** NH cancellation requires $m_1 c_{12}^2 c_{13}^2 + m_2 s_{12}^2 c_{13}^2 e^{i\alpha_{21}} \approx m_3 s_{13}^2 e^{i\alpha_{31}}$ — happens near $m_1 \sim 2$–$10$ meV with $\alpha_{21} \simeq \pi$.

**Exercise: Experimental sensitivity as a function of exposure**

The half-life sensitivity of a $\beta\beta_{0\nu}$ search depends on exposure $M \cdot t$,
background index (BI) and energy resolution $\Delta E$:

$$
T^{0\nu}_{1/2}{}^{\rm sens} \propto \begin{cases}
M \cdot t & \text{(zero-background)} \\
\sqrt{M \cdot t / (\mathrm{BI} \cdot \Delta E)} & \text{(background-dominated)}
\end{cases}
$$

**Questions:**

1. Compare the half-life reach of LEGEND-1000 vs nEXO vs NEXT-HD using their projected parameters.
2. At what exposure does each experiment transition from the zero-background to background-dominated regime?
3. Which parameter (mass, BI, or energy resolution) has the largest impact on the $m_{\beta\beta}$ reach?

```python
import majorana
majorana.exercise_sensitivity()
```

**Q2.** Transition to background-dominated when $N_{\rm bkg} \sim 1$, i.e. $M \cdot t \sim 1/(\text{BI} \cdot \Delta E)$. For typical next-gen targets ($\text{BI} \sim 10^{-5}$, $\Delta E \sim 5$ keV) this is $\sim 2 \times 10^{4}$ kg·yr — well above the 10-yr exposure, so next-gen runs essentially in the zero-background regime.

**Q3.** In the **zero-background** regime, **mass** (isotope mass × live time) drives the sensitivity linearly. In the **background-dominated** regime, $\text{BI}$ and $\Delta E$ enter as square roots — but both simultaneously matter. The largest lever is usually **BI reduction**, since it can span several orders of magnitude, whereas $\Delta E$ and $M$ scale less dramatically.
