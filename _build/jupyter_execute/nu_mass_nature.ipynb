## 1. Experimental Limits on the Neutrino Mass

### Neutrino mass from oscillations: recap

Oscillations $\Rightarrow$ mass-squared differences:

$$\Delta m^2_{21} = (7.41^{+0.21}_{-0.20})\times 10^{-5}\,\mathrm{eV}^2$$

$$|\Delta m^2_{31}| = (2.511^{+0.028}_{-0.027})\times 10^{-3}\,\mathrm{eV}^2 \quad (\text{NuFit-6.0, 2024})$$

Splittings fixed, absolute scale $m_1$ unknown. Two orderings:

| Ordering | $m_1$ | $m_2$ | $m_3$ | $\sum m_i$ (min) |
|---|---|---|---|---|
| **NH** | lightest | $m_1+7\times10^{-3}$ eV | $m_1+50\times10^{-3}$ eV | $\geq 57$ meV |
| **IH** | $m_3+50\times10^{-3}$ eV | $m_3+51\times10^{-3}$ eV | lightest | $\geq 100$ meV |

**Ordering status:**

 * NuFit-6.0: NH preferred at $\sim 2.5\sigma$
 * JUNO first results (2025): NH at $\sim 2\sigma$ from reactors
 * Definitive measurement needs full JUNO / DUNE / HyperK

### Kinematic neutrino mass measurements

$\beta$-decay endpoint spectroscopy — **model-independent** probe of the effective electron-antineutrino mass:

$$m^2_\nu \equiv \sum_i |U_{ei}|^2\, m_i^2$$

 * Independent of Majorana/Dirac, CP phases, cosmology
 * Shape distortion maximal near the endpoint $E_0$:

$$
\frac{dN}{dT} \propto (E_0 - T) \sqrt{(E_0 - T)^2 - m^2_\nu} \;\Theta(E_0 - T - m_\nu)
$$

## 2. How to Provide Mass to the Neutrinos

### Dirac neutrino masses

Dirac spinor — Dirac equation, decomposable into chiral parts:

$$\Psi = \Psi_L + \Psi_R$$

 * Global phase $\Psi' = e^{i\theta}\Psi$ $\Rightarrow$ conserved charge (electric, lepton number, …)
 * Mass term couples both chiralities:

$$-m \bar{\Psi} \Psi = -m (\bar{\Psi}_L \Psi_R + \bar{\Psi}_R \Psi_L)$$

**A mass term requires both left and right components.**

### Dirac masses in the SM

 * $L$ chiral fields: $SU(2)_L$ doublets; $R$ fields: singlets
 * Bare masses forbidden by gauge invariance $\Rightarrow$ arise via Yukawa + SSB

For one lepton family:

$$
L = \begin{pmatrix} \nu_L \\ l_L \end{pmatrix}, \quad l_R, \quad
\Phi = \begin{pmatrix} \phi^+ \\ \phi^0 \end{pmatrix}
\xrightarrow{\text{SSB}}
\langle \Phi \rangle = \begin{pmatrix} 0 \\ v/\sqrt{2} \end{pmatrix}
$$

Charged lepton: $\lambda_e \bar{L} \Phi \, e_R \Rightarrow m_e = \lambda_e \, v/\sqrt{2}$.

**Neutrinos remain massless** — no $\nu_R$ in the SM.

Need a **sterile right-chiral field** $\nu_R$, SM singlet, coupled via $\tilde{\Phi}$:

$$\tilde{\Phi} = i\sigma^2 \Phi^* \xrightarrow{\text{SSB}}
\langle \tilde{\Phi} \rangle = \begin{pmatrix} v/\sqrt{2} \\ 0 \end{pmatrix}$$

New Yukawa:

$$-\mathcal{L}_{\mathrm{mass},\,\nu} = \lambda_\nu \, \bar{L} \, \tilde{\Phi} \, \nu_R + \mathrm{h.c.} \quad\Rightarrow\quad m_\nu = \lambda_\nu \, v/\sqrt{2}$$

**Problem:** why is $m_\nu$ six orders of magnitude smaller than charged leptons?

$$\lambda_\nu \ll \lambda_l$$

No SM explanation — motivates the **Majorana** alternative.

### Majorana fermions

[E. Majorana](https://en.wikipedia.org/wiki/Ettore_Majorana) (1930s): neutral fermions can be **self-conjugate**.

| |
| :--: |
| <img src="./imgs/numass_majorana_abstract.png" width = 400> |

Majorana basis — all $\gamma^\mu$ purely imaginary:

$$
\gamma^0 = \begin{pmatrix} 0 & \sigma^2 \\ \sigma^2 & 0\end{pmatrix}, \;
\gamma^1 = \begin{pmatrix} i\sigma^3 & 0 \\ 0 & i\sigma^3\end{pmatrix}, \;
\gamma^2 = \begin{pmatrix} 0 & -\sigma^2 \\ \sigma^2 & 0\end{pmatrix}, \;
\gamma^3 = \begin{pmatrix} i\sigma^1 & 0 \\ 0 & i\sigma^1 \end{pmatrix}
$$

Dirac equation is real $\Rightarrow$ admits $\Psi(x) = \Psi^*(x)$: **particle = antiparticle**.

#### The Majorana condition

$$\Psi = \Psi^c, \quad \Psi^c \equiv C \bar{\Psi}^T = C \gamma^0 \Psi^*$$

$C$ obtained from covariance of the Dirac equation under charge conjugation.

In the Weyl-chiral basis:

$$
\Psi = \begin{pmatrix} \psi_L \\ \psi_R \end{pmatrix} \to
\Psi^c = i\gamma^2 \Psi^* = \begin{pmatrix} i\sigma^2 \psi_R^* \\ -i\sigma^2 \psi_L^* \end{pmatrix}
$$

Charge conjugate of a L spinor is R:

$$
\Psi_L = \begin{pmatrix} \psi_L \\ 0 \end{pmatrix} \to
(\Psi_L)^c = \begin{pmatrix} 0 \\ -i \sigma^2 \psi^*_L \end{pmatrix}
$$

#### Constructing the Majorana spinor

From a single L Weyl spinor $\psi_L$:

$$
\Psi^M = \Psi + \Psi^c = \begin{pmatrix} \psi_L \\ -i \sigma^2 \psi^*_L \end{pmatrix}
$$

**Properties of a Majorana fermion:**

 * Has L **and** R chiral components $\Rightarrow$ **has mass** (otherwise reduces to Weyl)
 * **No global phase symmetry** ($\psi_L \to e^{i\phi}\psi_L$ mixes signs in upper/lower blocks) $\Rightarrow$ **no conserved charge**
 * Purely neutral fermion

Neutrinos have mass — as demonstrated by oscillations — and they are the only fundamental fermions that could be Majorana particles.

*What, then, is the nature of neutrinos: Dirac or Majorana?*

### The Weinberg operator

Majorana mass for $\nu_L$:

$$-\frac{m}{2} (\overline{\nu_L} \nu^c_L + \overline{\nu^c_L} \nu_L)$$

Gauge invariance $\Rightarrow$ double Higgs coupling. The unique dim-5 operator [Wein]:

$$\mathcal{L}_W = -\frac{\alpha}{\Lambda} (\bar{L}\tilde{\Phi})(\tilde{\Phi}^T L^c) + \mathrm{h.c.}$$

After SSB:

$$m_\nu = \frac{\alpha v^2}{\Lambda}$$

For $\alpha \sim 1$, $v = 246$ GeV, $m_\nu \sim 1$ eV $\Rightarrow$ $\Lambda \sim 10^{14}$ GeV.

The Weinberg operator is the **only 5-dimensional operator** in the SM effective field theory expansion — it is the first signal of physics beyond the SM at low energy.

### Mixing and the $U_{PMNS}$ Matrix

With three families, the Yukawa couplings become $3\times3$ complex matrices.

The coupling with the Higgs allows Nature to relate flavour eigenstates to mass eigenstates via a unitary matrix $U$.

This is the matrix responsible, among other things, for neutrino oscillations.

The derivation depends on the neutrino nature: whether it is a Dirac or a Majorana fermion.

### Dirac masses: three families

Yukawa mixes flavours:

$$
-\mathcal{L}_H = \sum_{i,j=1}^3 \left(\lambda^l_{ij} \bar{L}'_i \Phi \, l'_{jR} + \lambda^\nu_{ij} \bar{L}'_i \tilde{\Phi} \, \nu'_{jR}\right) + \mathrm{h.c.}
$$

After SSB, diagonalise by two unitary transformations:

$$\frac{v}{\sqrt{2}} \lambda_\alpha = U^\dagger_\alpha \, M_\alpha V_\alpha$$

Weak ↔ mass eigenstates: $l_L = U_l l'_L$, $\nu_L = U_\nu \nu'_L$.

 * NC **unaffected** (unitarity)
 * CC acquires the **PMNS matrix**:

$$-\mathcal{L}_{CC} = \frac{g}{\sqrt{2}} \bar{l}_L \gamma^\mu (U^\dagger_l U_\nu) \nu_L W^+_\mu + \mathrm{h.c.}$$

Quark analogue: $V_{CKM}$.

### The $U_{PMNS}$ matrix (Pontecorvo–Maki–Nakagawa–Sakata) [MNS, Pon]

$$
U_{PMNS} = U^\dagger_l \, U_\nu
$$

The flavour and mass eigenstates are related by: $\nu'_\alpha = \sum_k U_{\alpha k} \, \nu_k$

**Dirac case**: $U_{PMNS}$ has 3 angles and **1 CP phase** ($\delta_{CP}$). We can reabsorb 5 phases using the global symmetries of the 6 Dirac fields.

### Majorana masses: three families

In the Majorana case, we extend the SM with the Weinberg operator for three families:

$$
-\mathcal{L}_m = \sum_{\beta, \beta'} \frac{\alpha^\nu_{\beta\beta'}}{\Lambda} (\bar{L}'_\beta \tilde{\Phi})(\tilde{\Phi}^T L'^c_{\beta'}) + \mathrm{h.c.}
$$

After SSB, the $3\times3$ symmetric complex matrix $\frac{v}{\sqrt{2}\Lambda}\alpha_\nu$ is diagonalised by a single unitary transformation.

The resulting mass matrix is symmetric — a hallmark of the Majorana nature.


**Majorana case**: Majorana fields have no global phase freedom, so only 3 phases (from charged leptons) can be reabsorbed. The $U_{PMNS}$ matrix has 3 angles and **3 phases**: $\delta_{CP}$ and two Majorana phases $\eta_1, \eta_2$.

The Majorana phases $\eta_{1,2}$ have no effect on oscillation probabilities — they can only be probed through lepton-number-violating processes such as $\beta\beta0\nu$.

### The seesaw mechanism

Weinberg operator is low-energy effective. UV completion at tree level $\Rightarrow$ **seesaw**.

| Type | New particle | Mass scale | Key feature |
|---|---|---|---|
| **Type I** | Heavy RH neutrino $N_R$ | $M_R \sim 10^{14}$ GeV | Minimal, most studied |
| **Type II** | Scalar triplet $\Delta$ | $M_\Delta \sim 10^{13}$ GeV | Same-sign dileptons |
| **Type III** | Fermion triplet $\Sigma$ | $M_\Sigma \sim 10^{13}$ GeV | Collider-testable |

#### Type I seesaw

Extend SM with $n$ heavy RH (sterile) neutrinos $N_{Ri}$, SM singlets.

$$-\mathcal{L} = y_\nu\,\bar{L}\,\tilde{\Phi}\,N_R + \frac{1}{2}M_R\,\overline{N^c_R}\,N_R + \text{h.c.}$$

After SSB ($m_D = y_\nu v/\sqrt{2}$, $M_R \gg m_D$):

$$
\mathcal{M} = \begin{pmatrix} 0 & m_D \\ m_D & M_R \end{pmatrix}
$$

Diagonalisation:

$$m_\nu \approx \frac{m_D^2}{M_R} \;(\text{light}), \qquad M_N \approx M_R \;(\text{heavy})$$

Larger $M_R$ $\Rightarrow$ smaller $m_\nu$ — hence *seesaw*.

 * $N_R$ can drive matter-antimatter asymmetry via **leptogenesis**

## 3. How to Determine the Nature of the Neutrino

### Signature of Dirac and Majorana neutrinos

| |
| :--: |
| <img src="./imgs/pidecay_majorana_dirac.png" width = 600> |

**How to distinguish Majorana vs Dirac?** If Majorana, a $\nu$ from $\pi^+$ could produce either $\mu^+$ or $\mu^-$ — LNV.

**SM massless $\nu$: triple degeneracy**

|  |  |  |  |
| :--:  | :--: | :--: | :--: |
| $\nu$ | L chirality | $(-)$ helicity | $L_\nu = +1$ |
| $\bar{\nu}$ | R chirality | $(+)$ helicity | $L_\nu = -1$ |

Chirality, helicity and $L_\nu$ all equivalent. With mass, degeneracy breaks.

If Majorana, LNV amplitude is helicity-suppressed. Probabilities:

| | $h = -1$ | $h = +1$ |
| :--:  | :--: | :--: |
| $\nu_L$ | $\sim 1$ | $\mathcal{O}(m/E)$ |
| $\nu_R$ | $\mathcal{O}(m/E)$ | $\sim 1$ |

**Devastating suppression:**

 * $m_\nu \sim 1$ eV, $E \sim 1$ MeV $\Rightarrow$ $\sim 10^{-12}$
 * $m_\nu \sim 1$ eV, $E \sim 1$ GeV $\Rightarrow$ $\sim 10^{-18}$

LNV direct observation in scattering is impossible. Majorana and Dirac are *practically indistinguishable* — except via the rare $\beta\beta0\nu$.

### Double beta decay without neutrinos

| |
| :--: |
| <img src="./imgs/bb0nu_Feynman.png" width = 400> |

Only experimental probe of the Majorana nature:

$$(A, Z) \to (A, Z+2) + 2e^- + Q_{\beta\beta}$$

All three $\nu$ participate, each suppressed by $U^2_{ei} \, m_i / E$. Amplitude $\propto$ **effective Majorana mass**:

$$m_{\beta\beta} \equiv \left|\sum_i U^2_{ei} \, m_i\right|$$

### $\beta\beta 2\nu$: the allowed double beta decay

Proposed by [M. Goeppert-Mayer](https://en.wikipedia.org/wiki/Maria_Goeppert-Mayer) (1935):

$$(A, Z) \to (A, Z+ 2) + 2 e^- + 2 \bar{\nu}_e + Q_{\beta\beta}$$

 * 2nd-order weak decay — occurs in 35 isotopes where single $\beta$ is forbidden
 * Relevant isotopes: $^{48}$Ca, $^{76}$Ge, $^{82}$Se, $^{100}$Mo, $^{130}$Te, $^{136}$Xe
 * $T_{1/2} \sim \mathcal{O}(10^{18}$–$10^{24})$ yr

### The effective Majorana mass

Depends on $U_{\alpha i}$, lightest mass $m_0$, Majorana phases.

**NH:**

$$
m_{\beta\beta} = \left| m_0 c^2_{12} c^2_{13} + \sqrt{m^2_0 + \Delta m^2_{21}} \, s^2_{12} c^2_{13} e^{2i (\eta_2 -\eta_1)} + \sqrt{m^2_0 + \Delta m^2_{32} + \Delta m^2_{21}} \, s^2_{13} e^{-2i(\delta_{CP} + \eta_1)}\right|
$$

**IH:**

$$
m_{\beta\beta} = \left| m_0 s^2_{13} + \sqrt{|\Delta m^2_{32}| + m^2_0} \, c^2_{13} \left| s^2_{12} + c^2_{12} \, e^{2i(\eta_2 - \eta_1)} \right| \right|
$$

For $m_0 \to 0$: $m_{\beta\beta} \simeq 1.1$–$4.2$ meV (NH), $15$–$50$ meV (IH).

### Experimental signature of $\beta\beta0\nu$

 * Signal: **peak** at $Q_{\beta\beta}$ (both $e^-$ carry all the energy)
 * $\beta\beta2\nu$: continuous spectrum (missing energy in neutrinos)

Expected events:

$$N_{\beta\beta} = \ln(2) \, \frac{N_A}{W} \, \frac{a \, \epsilon \, M \, t}{T^{0\nu}_{1/2}}$$

with $N_A$ Avogadro, $W$ molar mass, $a$ abundance, $\epsilon$ RoI efficiency, $M$ target mass, $t$ live time.

### Background sources and mitigation

**Signal identification:**

 * **Energy resolution** — defines RoI around $Q_{\beta\beta}$, $\Delta E = \mathrm{FWHM} \simeq 2.355\,\sigma$
 * **Topology** (2-$\beta$ tracks), pulse-shape analysis, fiducialisation

**Main backgrounds:**

 * **Cosmogenic muons + spallation** — mitigated by depth + active vetoes, reduction $\mathcal{O}(10^6)$
 * **Natural radioactivity** ($^{238}$U, $^{232}$Th chains) — $^{208}$Tl, $^{214}$Bi $\gamma$'s near $Q_{\beta\beta}$
 * Ultra-radiopure materials required: $\lesssim 10^{-6}$ Bq/kg

### Sensitivity regimes

Background events in RoI: $N_{\mathrm{bkg}} \propto b \, \Delta E \, M \, t$.

**Two regimes** for the half-life sensitivity $\mathcal{S}(T^{0\nu}_{1/2})$:

 * **Background-free** ($N_{\mathrm{bkg}} \ll 1$):

$$\mathcal{S}(T^{0\nu}_{1/2}) \propto \epsilon \, M \, t$$

 * **Background-dominated** ($N_{\mathrm{bkg}} \gg 1$):

$$\mathcal{S}(T^{0\nu}_{1/2}) \propto \sqrt{\frac{M \, t}{b \, \Delta E}}$$

| |
| :--: |
| <img src="./imgs/Xe_mbb_vs_exposure_bkginRoI.png" width = 400 align='center'> |

Projected $m_{\beta\beta}$ limits for an ideal $^{136}$Xe experiment vs exposure, for different BI in the RoI [[4]](https://arxiv.org/abs/1502.00581).

Target for IH coverage: tonne-scale + $\lesssim 1$ count/(RoI·ton·yr) — the holy grail.

### Detector requirements

The main ingredients for a successful $\beta\beta0\nu$ detector are:

   - **Large exposure** of the target isotope: $M \cdot t \sim \mathcal{O}(1)$ ton yr
   - **Excellent energy resolution**: $\Delta E \sim \mathcal{O}(1\text{--}0.1)$% FWHM at $Q_{\beta\beta}$
   - **Ultra-low background**: $b \sim \mathcal{O}(0.1)$ counts/(RoI ton yr), requiring radiopure materials, underground location, and active background rejection

Different isotopes and detection techniques offer complementary trade-offs between these three requirements, as we shall see in the next section.

## The Experimental Search for $\beta\beta0\nu$ Decays

### Liquid xenon experiments

$^{136}$Xe ($Q_{\beta\beta} = 2458$ keV): large homogeneous target, scalable to tonne.

**Two techniques:**

 * **Liquid Xe TPC** — charge + light readout — EXO-200
 * **Xe in liquid scintillator** — KamLAND-Zen

**Why $^{136}$Xe:**

 * High $Q_{\beta\beta}$, above most $\gamma$ backgrounds
 * 8.9% natural abundance, enrichable $> 80\%$
 * Self-shielding, scalable to tonnes

**Key backgrounds:**

 * $^{214}$Bi ($2448$ keV $\gamma$) — 10 keV below $Q_{\beta\beta}$, most dangerous
 * $^{208}$Tl ($2615$ keV $\gamma$) — Compton tail in RoI
 * $\beta\beta2\nu$ tail (limited by resolution)
 * Cosmogenic $^{137}$Xe ($T_{1/2} = 3.8$ min)
 * Radon emanation

### EXO-200 (2011–2015)

 * [WIPP](https://www.wipp.energy.gov), New Mexico (1600 m.w.e.)
 * 175 kg of $^{136}$Xe enriched to 80%
 * Symmetric LXe TPC: wires for ionization, APDs for scintillation
 * HFE-7000 bath, lead castle, active $\mu$ veto

#### XLZD

Next-gen liquid-xenon TPC — merger of LZ + XENON + DARWIN.
Dual-purpose: **dark matter + $0\nu\beta\beta$** [[XLZD-DB]](https://arxiv.org/abs/2410.17137).

 * **60–80 t of natural Xe** (8.9% $^{136}$Xe $\Rightarrow$ ~5.3–7.1 t of $^{136}$Xe — no enrichment)
 * Dual-phase LXe TPC with SiPM readout, $\sigma_E \sim 0.8\%$ at $Q_{\beta\beta}$
 * Host lab under selection (SURF / LNGS / Boulby / SNOLAB)
 * Status (2025): **Design Book published**; first data early 2030s, 10-yr run

**$0\nu\beta\beta$ sensitivity (10 yr)** [[XLZD-bb]](https://arxiv.org/abs/2410.19016):

 * $3\sigma$ discovery: $T^{0\nu}_{1/2} > 5.7 \times 10^{27}$ yr
 * 90% CL exclusion: $T^{0\nu}_{1/2} > 1.3 \times 10^{28}$ yr
 * $m_{\beta\beta}$: 4.8–20.5 meV (exclusion), 7.3–31.3 meV ($3\sigma$)
 * **Excludes IH, starts probing NH** for most NMEs

### CUORE (2017–2024)

Cryogenic bolometric search for $0\nu\beta\beta$ of $^{130}$Te at LNGS.

 * 988 TeO$_2$ crystals (19 towers), ~206 kg of $^{130}$Te, operated at **~10 mK**
 * Natural Te (34% $^{130}$Te) — no enrichment
 * $Q_{\beta\beta} = 2527$ keV, $\Delta E \sim 7$ keV FWHM
 * BI $\sim 1.4 \times 10^{-2}$ cts/(keV·kg·yr) — **background-limited**

**Final result — 2 tonne·yr** (Science 2024, [arXiv:2404.04453](https://arxiv.org/abs/2404.04453)) [10]:

 * $T^{0\nu}_{1/2}(^{130}\text{Te}) > 3.5 \times 10^{25}$ yr (90% CI, Bayesian)
 * $m_{\beta\beta} \lesssim 70$–$240$ meV
 * Sensitivity gain needs particle ID $\Rightarrow$ CUPID

#### CUPID — CUORE upgrade

Scintillating Li$_2$$^{100}$MoO$_4$ bolometers + Ge light detectors in the CUORE cryostat at LNGS [[CUPID]](https://arxiv.org/abs/2203.08386).

 * 1596 Li$_2$MoO$_4$ crystals, $\geq 95\%$ enriched in $^{100}$Mo (~240 kg of $^{100}$Mo)
 * $Q_{\beta\beta}(^{100}\text{Mo}) = 3034$ keV — above most $\gamma$ backgrounds
 * Heat + light $\Rightarrow$ rejects $\alpha$ background by ~100×
 * Target BI $\sim 10^{-4}$ cts/(keV·kg·yr), $\Delta E \sim 5$ keV FWHM

**Sensitivity:**

 * 90% CL exclusion: $T^{0\nu}_{1/2} > 1.8 \times 10^{27}$ yr
 * $3\sigma$ discovery: $\sim 10^{27}$ yr
 * $m_{\beta\beta} \sim 9$–$15$ meV — covers full IH
 * Data-taking: end of the decade

### High-pressure gaseous xenon experiments

Gaseous $^{136}$Xe at high pressure (10–15 bar) combines excellent energy resolution ($<1$% FWHM) with a unique **topological signature**: the two $\beta$-electrons produce tracks with energy depositions (blobs) at both ends, providing powerful background rejection.

**Key observations:**

 * KamLAND-Zen currently leads in $m_{\beta\beta}$ due to its very large $^{136}$Xe mass, despite modest energy resolution ($\sim 10\%$ FWHM).
 * LEGEND delivers the best half-life limit per unit exposure thanks to $\sim 0.1\%$ FWHM resolution and BI below $10^{-3}$.
 * CUORE is background-limited ($\text{BI} \sim 10^{-2}$); the discovery gain of CUPID comes from scintillation particle ID, reducing the $\alpha$ background by two orders of magnitude.
 * All next-generation experiments aim at $\text{BI} \lesssim 10^{-4}$ to approach the zero-background regime at their planned exposures.
 * The ranges in $m_{\beta\beta}$ reflect the spread of nuclear matrix elements; the isotope diversity is essential to break this degeneracy if a signal is found.

**Exercise: Experimental sensitivity as a function of exposure**

The half-life sensitivity of a $\beta\beta_{0\nu}$ search depends on exposure $M \cdot t$,
background index (BI) and energy resolution $\Delta E$:

$$
T^{0\nu}_{1/2}{}^{\rm sens} \propto \begin{cases}
M \cdot t & \text{(zero-background)} \\
\sqrt{M \cdot t / (\mathrm{BI} \cdot \Delta E)} & \text{(background-dominated)}
\end{cases}
$$

1. Compare the half-life reach of LEGEND-1000 vs nEXO vs NEXT-HD using their projected parameters.
2. At what exposure does each experiment transition from the zero-background to background-dominated regime?
3. Which parameter (mass, BI, or energy resolution) has the largest impact on the $m_{\beta\beta}$ reach?

## Conclusions

**Measuring $m_\nu$ is one of the most urgent tasks in fundamental physics.**

 * Oscillations: $\Sigma m_\nu^{\min}(\mathrm{NH}) \approx 59$ meV, $\approx 100$ meV (IH)
 * KATRIN: $m^{\mathrm{eff}}_{\nu_e} < 0.45$ eV (goal $< 0.2$ eV)
 * Cosmology (Planck + DESI): $\Sigma m_\nu < 0.072$–$0.12$ eV
 * **Tension** cosmology vs. oscillations $\Rightarrow$ motivates model-independent lab measurements

**Majorana nature is experimentally very hard.**

 * Needs BI $\lesssim 10^{-4}$ cts/(keV·kg·yr), tonne-scale isotopic mass, high resolution, radiopurity, efficiency
 * Current: $T^{0\nu}_{1/2} > 3.8 \times 10^{26}$ yr (KamLAND-Zen), $> 1.9 \times 10^{26}$ yr (LEGEND-200 + GERDA + MAJORANA)
 * Next gen (LEGEND-1000, CUPID, KamLAND2-Zen, NEXT-HD, XLZD, nEXO) **covers IH** ($m_{\beta\beta} \gtrsim 15$ meV)
 * **NH still out of reach** — requires even more ambitious facilities

**$m_\nu$ connects Higgs physics and new physics at high scale.**

 * SM needs new physics: Dirac (tiny Yukawa $\sim 10^{-12}$) or Majorana (Weinberg op. at $\Lambda \sim 10^{14}$ GeV)
 * Majorana $\Rightarrow$ seesaw + leptogenesis + baryon asymmetry
 * Forces us to re-examine the Higgs mechanism for neutrinos

**Neutrinos are a fundamental component of the Universe.**

 * C$\nu$B number density: $n_\nu \simeq 336$ cm$^{-3}$ — 2nd most abundant after photons ($n_\gamma \simeq 411$ cm$^{-3}$)
 * $\Omega_\nu h^2 = \Sigma m_\nu / 93.14\,\text{eV}$
 * Minimum ($\Sigma m_\nu \simeq 0.06$ eV): $\Omega_\nu \simeq 1.4 \times 10^{-3}$ ($\sim 0.14\%$)
 * Current cap: $\Omega_\nu < 3 \times 10^{-3}$ ($\sim 0.5$–$1\%$ of $\Omega_m$)
 * Key role in large-scale structure

Pinning down $m_\nu$ and its Dirac/Majorana nature is about **the origin of mass, the matter-antimatter asymmetry, and the structure of the Universe**.

## Appendix: About heavy neutral leptons

### Why heavy neutrinos?

Type-I seesaw: add $n \geq 2$ right-handed (sterile) $N_R$, SM singlets. Mass matrix

$$\mathcal{M}_\nu = \begin{pmatrix} 0 & m_D \\ m_D^{T} & M_R \end{pmatrix} \Rightarrow m_\nu \simeq -m_D M_R^{-1} m_D^{T}, \quad M_N \simeq M_R$$

Active-sterile mixing:

$$|U_{\alpha N}|^{2} \simeq m_\nu / M_N$$

**Why look for HNLs?**

 * Direct evidence of the **seesaw mechanism**
 * $\nu$MSM (keV–GeV): explains $m_\nu$ + **leptogenesis** + dark matter
 * **Long-lived** for $M_N \lesssim m_W$ $\Rightarrow$ displaced-vertex signature
 * If Majorana, contribute to $m_{\beta\beta}$ in $0\nu\beta\beta$

Parameter space: plane $(M_N, |U_{\alpha N}|^{2})$, spanning many decades — no single experiment covers it all.

### HNL searches at the LHC

ATLAS, CMS, LHCb in $W \to \ell N$, $Z \to \nu N$ and heavy-flavour decays [[PDG-HNL]](https://pdg.lbl.gov/2025/listings/rpp2025-list-heavy-neutral-leptons.pdf).

**Mass ranges:**

 * $M_N \lesssim m_B$: LHCb, $B \to X \ell N$, $N \to \ell \pi$ displaced in VELO [[LHCb-HNL]](https://arxiv.org/abs/2512.14551)
 * $m_B \lesssim M_N \lesssim m_W$: ATLAS/CMS, $W \to \ell N$, displaced vertices; CMS extends to muon system for long $c\tau$
 * $M_N \gtrsim m_W$: prompt same-sign $\ell^{\pm}\ell^{\pm} jj$ — LNV signature unique to Majorana

**Status (Run 2 + partial Run 3):**

 * $|U_{\alpha N}|^{2} \lesssim 10^{-6}$–$10^{-7}$ at $M_N \sim 5$–$30$ GeV
 * HL-LHC projection: gain ~1 order of magnitude
 * Complementary to beam dumps below $m_B$

### SHiP — CERN SPS Beam Dump Facility

Dedicated beam-dump for **feebly-interacting particles** at CERN [[SHiP25]](https://arxiv.org/abs/2504.06692).

**Concept:**

 * 400 GeV protons on thick target $\Rightarrow$ $\sim 10^{20}$ charm + beauty hadrons in 15 yr
 * $D, B \to \ell N \Rightarrow$ displaced 2-track vertex ($N \to \mu\pi$, $\mu\ell\nu$) in evacuated 50 m decay volume
 * Background-free thanks to upstream absorber + muon shield

**Reach:**

 * $M_N \sim 0.5$–$5$ GeV, $|U_{\alpha N}|^{2}$ down to $\sim 10^{-10}$
 * Covers large portion of $\nu$MSM cosmological band
 * Also probes dark photons, dark scalars, ALPs, LDM

**Status (2025):**

 * TDR green light **2024**; ECN3 / BDF construction endorsed
 * Funding applications in Italy, Switzerland, UK, etc.
 * First beam **2031**, 15-yr run

### Other long-lived particle experiments

Complementary LLP detectors at the LHC [[LLP-rev]](https://arxiv.org/abs/2512.13011):

| Experiment | Host | Status | Mass reach | Concept |
|:--|:--:|:--:|:--:|:--|
| **FASER / FASER$\nu$** | LHC IP1 (480 m) | Running (Run 3) | $\lesssim 1$ GeV | Forward decay volume along beam line |
| **FASER2** | HL-LHC FPF | Proposed | $\lesssim 5$ GeV | 10× larger, dedicated HL-LHC facility |
| **CODEX-b** [[CODEX]](https://arxiv.org/abs/2505.05952) | LHCb cavern | Proposed (demonstrator) | $\lesssim 10$ GeV | $10^3$ m$^3$ shielded box, transverse to beam |
| **MATHUSLA** | Surface above CMS | Proposed | $\lesssim 30$ GeV | $\sim 100\times100$ m$^2$ tracker, $c\tau \sim 100$ m–km |
| **ANUBIS** | Shaft above ATLAS | Proposed | $\lesssim 30$ GeV | Tracker in PX14 shaft — low-cost MATHUSLA alternative |

**Complementarity:**

 * Sub-GeV: SHiP, FASER
 * Few–tens GeV: CODEX-b, MATHUSLA, ANUBIS
 * Up to TeV: ATLAS/CMS prompt

If HNLs in the 0.1–10 GeV range are discovered **and** are Majorana, they contribute directly to $m_{\beta\beta}$ — linking collider searches with the tonne-scale $0\nu\beta\beta$ programme.
